### Data ingestion

```python
import xml.etree.ElementTree as ET
import pandas as pd


tree = ET.parse('Dataset/customers.xml')

root = tree.getroot()

list_customers = []

for customer in root.findall('customer'):
   data = {
      'customer_id':  customer.find('customer_id').text, 
      'first_name': customer.find('first_name').text, 
      'last_name': customer.find('last_name').text, 
      'email': customer.find('email').text,
      'city': customer.find('city').text
   }

   list_customers.append(data)
df_customer  = pd.DataFrame(list_customers)
df_customer.head()    
    
```

```python
import xml.etree.ElementTree as ET
import pandas as pd


tree = ET.parse('Dataset/products.xml')

root = tree.getroot()

list_products = []

for product in root.findall('product'):
   data = {
      'product_id':  product.find('product_id').text, 
      'product_name': product.find('product_name').text, 
      'category': product.find('category').text, 
      'brand': product.find('brand').text,
      'price': product.find('price').text
   }

   list_products.append(data)
df_products  = pd.DataFrame(list_products)
df_products.head()    
    
```

```python
import xml.etree.ElementTree as ET
import pandas as pd


tree = ET.parse('Dataset/orders.xml')

root = tree.getroot()

list_orders = []

for order in root.findall('order'):
   data = {
      'order_id':  order.find('order_id').text, 
      'customer_id': order.find('customer_id').text, 
      'product_id': order.find('product_id').text, 
      'quantity': order.find('quantity').text,
      'order_date': order.find('order_date').text
   }

   list_orders.append(data)
df_orders  = pd.DataFrame(list_orders)
df_orders.head()    
    
```

### Silver layer


```python
from sqlalchemy import create_engine

engine = create_engine(
    "mssql+pyodbc://localhost/Sales_dwh"
    "?trusted_connection=yes"
    "&driver=ODBC+Driver+17+for+SQL+Server"
    "&TrustServerCertificate=yes"
)
```

```python
df_customers =pd.read_sql("select * from bronze1.Customers",engine)
df_products =pd.read_sql("select * from bronze1.Products",engine)
df_orders =pd.read_sql("select * from bronze1.Orders",engine)

```

```python
df_customers.dtypes
```

```python
df_customers['customer_id']  = df_customers['customer_id'].astype(int)
```

```python
df_customers.dtypes
```

```python
df_customers['full_name']  = df_customers['firstname'] + ' ' + df_customers['lastname']
```

```python
df_customers.drop(columns=['firstname', 'lastname'], inplace=True)
```

```python
df_products.head()
```

```python
df_products['product_id']  = df_products['product_id'].astype(int)
df_products['price']  = df_products['price'].astype(float)
```

```python
df_products.dtypes
```

```python
df_orders.head()
```

```python
df_orders.dtypes
```

```python
df_orders['Order_id']  = df_orders['Order_id'].astype(int)
df_orders['customer_id'] = df_orders['customer_id'].astype('Int64')
df_orders['product_id']  = df_orders['product_id'].astype(int)
df_orders['quantity']  = df_orders['quantity'].astype(int)  
df_orders['order_date'] = pd.to_datetime(df_orders['order_date'], format='%Y-%m-%d')
```

```python
df_customers
df_products
df_orders
```

```python
df_customers.to_sql(
    'customers',
    con = engine,
    schema = 'silver',
    if_exists = 'append',
    index = False
)
```

```python
df_products.to_sql(
    'products',
    con = engine,
    schema = 'silver',
    if_exists = 'append',
    index = False
)
```

```python
df_orders.to_sql(
    'orders',
    con = engine,
    schema = 'silver',
    if_exists = 'append',
    index = False
)
```

### Gold layer

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mssql+pyodbc://localhost/Sales_dwh"
    "?trusted_connection=yes"
    "&driver=ODBC+Driver+17+for+SQL+Server"
    "&TrustServerCertificate=yes"
)
```

```python
df_custmers = pd.read_sql("select * from silver.customers", engine)
df_products = pd.read_sql("select * from silver.products", engine)
df_orders = pd.read_sql("select * from silver.orders", engine)
```

```python
df_output = pd.merge(df_orders, df_custmers, how='inner', on='customer_id')
```

```python
df_output
```

```python
df_output = pd.merge(df_output, df_products, how='inner', on='product_id')
```

```python
df_output.head()
```

```python
df_output['total_amount'] = df_output['quantity'] * df_output['price']
```

```python
df_output.dtypes
```

```python
df_output.to_sql(
    'main',
    con = engine,
    schema = 'gold',
    if_exists = 'append',
    index = False
)
```
